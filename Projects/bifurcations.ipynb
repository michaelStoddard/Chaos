# Change of Base

# Codomain 1

## Saddle-Node (fold)

Normal Form
$$\boxed{\dot x = \mu\mp x^2}$$

Necessary conditions
$$\begin{align}&\begin{cases}f|_0=0\\
f_x|_0=0\end{cases}\\
&f_{\mu}|_0\neq  0\\
&f_{x^2}|_0\neq  0\end{align}$$ 
Where $(1)$ is the condition for a nonhyperbolic fixed point. $(2)$ and $(3)$ are the partial derivatives with respect to $\mu$ and the second partial with repect to $x$, respectivly. All evaluated at the orgin $(x,\mu)=(0,0)$

## Transcritical

Normal Form
$$\boxed{\dot x = \mu x \mp x^2}$$

Necessary conditions
$$\begin{align}&\begin{cases}f|_0=0\\
f_x|_0=0\end{cases}\\
&f_{\mu}|_0     =  0\\
&f_{x\mu}|_0 \neq  0\\
&f_{x^2}|_0  \neq  0\end{align}$$ 
Where $(1)$ is the condition for a nonhyperbolic fixed point. $(2)$ through $(4)$ are the partial derivatives. All evaluated at the orgin $(x,\mu)=(0,0)$.

## Pitchfork

Normal Form
$$\boxed{\dot x = \mu x \mp x^3}$$
with $-x^3$ and $+x^3$ corresponding to super/sub critical bifurcations respectivly. Moreover, the former is associated with the origin loosing stability, and the latter having stability.

Necessary conditions
$$\begin{align}&\begin{cases}f|_0=0\\
f_x|_0=0\end{cases}\\
&f_{\mu}|_0     = 0\\
&f_{x^2}|_0     = 0\\
&f_{x\mu}|_0 \neq 0\\
&f_{x^3}|_0  \neq 0\end{align}$$ 
Where $(1)$ is the condition for a nonhyperbolic fixed point. $(2)$ through $(5)$ are the partial derivatives. All evaluated at the orgin $(x,\mu)=(0,0)$.
