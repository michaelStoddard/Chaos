## Autonomous System (2D)

```python
x1, x2, mu= smp.symbols(r'x_1 x_2 \mu')

dx1 = x2
dx2 = -x1 + mu*x2*(1-x1**2) 
```

```python

Jac = smp.Matrix([[smp.diff(dx1,x1),smp.diff(dx1,x2)],
                  [smp.diff(dx2,x1),smp.diff(dx2,x2)]])

Jac
```

```python
J0 = Jac.subs([[x1,0],[x2,0],[mu,1]])
J0
```

```python
J0.eigenvects()
```

```python
ev1 = J0.eigenvects()[0][2][0]
ev2 = J0.eigenvects()[1][2][0]
ev1
```

```python

P = smp.Matrix([[ev1.T],
               [ev2.T]]).T
P
```

```python

s, u = smp.symbols('s u')

P_inv = P.inv()
su = smp.Matrix([s,u])
xy_dot = smp.Matrix([dx1,dx2])
```

```python
trans = P@su
trans
```

```python
su_dot = P_inv@xy_dot
su_dot
```

```python
su_dot = su_dot.subs([[x1,trans[0]],
                       [x2,trans[1]]])
```

```python
ds =  su_dot[0].expand()
du = su_dot[1].expand()
```

```python
A = ds.coeff(s,1).coeff(u,0).coeff(mu,0)
A
```

```python
g1 = ds - A*s
g1
```

```python
B = du.coeff(s,0).coeff(u,1).coeff(mu,0)
B
```

```python
g2 = du - B*u

g2
```

```python
k1, k2, k3 = smp.symbols(r'\kappa_1 \kappa_2 \kappa_3')

h = k1*s**2 + k2*s*mu + k3*mu**2

Dh = smp.diff(h,s)
```

```python
N = Dh*(A*s + g1) - B*h - g2
N = N.subs(u,h).expand()
N
```

```python
N.coeff(s,0).coeff(mu,0)
```

```python
N.coeff(s,1).coeff(mu,0)
```

```python
N.coeff(s,2).coeff(mu,0)
```

```python
N.coeff(s,0).coeff(mu,1)
```

```python
N.coeff(s,0).coeff(mu,2)
```

```python
N.coeff(s,1).coeff(mu,1)
```

# Numerical solution

```python
import numpy as np

from scipy.integrate import odeint
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

```python
clr = ['cornflowerblue',
      'gold',
      'tomato',
      'dimgray']
```

Eigenvalue pair

$$\frac{\mu}{2} \pm \frac{\sqrt{(\mu - 2)*(\mu + 2)}}{2}$$

When $\mu=1$

$$\frac{1}{2} \pm \frac{\sqrt{3}}{2}\mathcal{i}$$

```python
class vanDerPol:

    clr = ['cornflowerblue', 'gold', 'tomato', 'dimgray']

    def __init__(self, mu, Af, wf):
        self.mu = mu
        self.Af = Af
        self.wf = wf
        self.soln = None
        self.T = None
        self.limC = None
        self.poinSec = None

    def drive(self, t):
        return self.Af * np.cos(self.wf * t)

    def dif_Eq(self, X, t):
        x1, x2 = X
        dx1 = x2
        dx2 = self.mu * x2 * (1 - x1**2) - x1 + self.drive(t)
        return [dx1, dx2]

    def simulate(self, x0, t):
        S = odeint(self.dif_Eq, x0, t).T
        self.soln = [S, t]

    def get_limCycle(self):
        x, y = self.soln[0]
        t = self.soln[1]
        dx = np.diff(x)
        idx = np.where(dx <= 0)[0]
        jump = np.where(np.diff(idx) > 1)[0]
        idx = idx[jump + 1]
        theta0 = idx[-3]
        thetaf = idx[-2]
        self.T = t[thetaf] - t[theta0]
        self.limC = [x[theta0:thetaf], y[theta0:thetaf]]

    def comp_poincar(self):
        x, y = self.soln[0]
        t = self.soln[1]
        diff = np.diff(x**2)
        idx = np.where(diff >= 0)[0]
        idx = idx[np.where(np.diff(idx) > 1)[0] + 1]
        pos = np.where(y[idx] > 0)[0]
        idx = idx[pos]
        t_real  = t[idx]
        t_phase = (t_real * self.wf) % (2 * np.pi)
        self.poinSec = [t_phase, y[idx], x[idx], t_real]
        return self.poinSec

    def _ss_start_time(self, n_tail=30, tol=2.0):
        """Return simulation time where Poincaré data has converged to its attractor."""
        if self.poinSec is None:
            self.comp_poincar()
        _, y_map, x_map, t_real = self.poinSec
        pts = np.column_stack([x_map, y_map])
        n = max(2, min(n_tail, len(pts) // 4))
        tail   = pts[-n:]
        center = tail.mean(axis=0)
        radius = np.max(np.linalg.norm(tail - center, axis=1)) * tol
        dists  = np.linalg.norm(pts - center, axis=1)
        outside = np.where(dists > radius)[0]
        if len(outside) == 0:
            return t_real[0]
        last = outside[-1]
        return t_real[last + 1] if last + 1 < len(t_real) else t_real[-1]

    def plt_sim(self):
        x, y = self.soln[0]
        t = self.soln[1]
        fig = plt.figure(figsize=(14, 6))
        gs  = gridspec.GridSpec(1, 4)
        p1  = fig.add_subplot(gs[:3])
        p2  = fig.add_subplot(gs[3:])
        p1.plot(t, x, color=self.clr[0], label=r'$x_1$')
        p1.plot(t, y, color=self.clr[1], label=r'$x_2$')
        p1.plot(t, self.drive(t), color=self.clr[2], label=r'$F(t)$')
        p1.grid()
        p1.legend()
        p2.plot(x, y)
        p2.grid()
        p2.yaxis.set_label_position("right")
        p2.yaxis.tick_right()
        p2.set_xlabel(r'$x_1$')
        p2.set_ylabel(r'$x_2$')

    def plt_attractor(self, n_tail=30, tol=2.0):
        """Phase portrait and Poincaré section side by side, transients in gray."""
        if self.poinSec is None:
            self.comp_poincar()
        t_phase, y_map, x_map, t_real = self.poinSec
        x, y = self.soln[0]
        t    = self.soln[1]

        t_ss      = self._ss_start_time(n_tail=n_tail, tol=tol)
        i_ss      = np.searchsorted(t, t_ss)
        i_poin_ss = np.searchsorted(t_real, t_ss)

        fig, axes = plt.subplots(1, 2, figsize=(12, 5))

        # Phase portrait
        axes[0].plot(x[:i_ss], y[:i_ss], color='silver', lw=0.7, label='transient')
        axes[0].plot(x[i_ss:], y[i_ss:], color=self.clr[0], lw=1.2, label='steady state')
        axes[0].set_xlabel(r'$x_1$')
        axes[0].set_ylabel(r'$x_2$')
        axes[0].set_title('Phase portrait')
        axes[0].legend()
        axes[0].grid()

        # Poincaré section
        if i_poin_ss > 0:
            axes[1].scatter(x_map[:i_poin_ss], y_map[:i_poin_ss],
                            c='silver', s=15, zorder=2, label='transient')
        axes[1].scatter(x_map[i_poin_ss:], y_map[i_poin_ss:],
                        c=self.clr[2], s=20, zorder=3, label='steady state')
        axes[1].set_xlabel(r'$x_1$')
        axes[1].set_ylabel(r'$x_2$')
        axes[1].set_title('Poincaré section')
        axes[1].legend()
        axes[1].grid()

        plt.suptitle(rf'Van der Pol  $\mu={self.mu}$,  $A={self.Af}$,  $\omega={self.wf}$')
        plt.tight_layout()
```

## Method 2

```python
def vanDerPol(X,mu):
        x1, x2 = X

        dx1 = x2
        dx2 = mu*x2*(1-x1**2)-x1 #self.Af*np.sin(self.wf*t)

        return [dx1,dx2]
```

```python
x = np.arange(-4,4,0.1)
dx = np.arange(-5,5,0.1)

xG, dxG = np.meshgrid(x,dx)
```

```python
def vanDerPol_orb(X, t, mu, Af, wf):
    x1, x2 = X
    dx1 = x2
    dx2 = mu*x2*(1 - x1**2) - x1 + Af*np.sin(wf*t)
    return [dx1, dx2]
```

```python
x_orb  = np.arange(-2,2,0.5)
dx_orb = np.arange(-3,3,0.5)
t      = np.linspace(0,20,1000)

field = np.meshgrid(x_orb,dx_orb)
```

```python
X0_x = field[0].flatten()
X0_y = field[1].flatten()
```

```python
mu_val = 1
Af_val = 1.2
wf_val = 1.2

orbs = np.empty((X0_x.size, 2, t.size))

for i in range(X0_x.size):
    x = X0_x[i]
    y = X0_y[i]
    orbs[i] = odeint(vanDerPol_orb, [x, y], t, args=(mu_val, Af_val, wf_val)).T
```

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

def get_field(t_val):
    """Forced Van der Pol vector field at time t_val."""
    fdx1 = dxG
    fdx2 = mu_val*dxG*(1 - xG**2) - xG + Af_val*np.sin(wf_val*t_val)
    return fdx1, fdx2

fig, ax = plt.subplots(figsize=(7, 6))

n_frames = orbs.shape[2]
step = max(1, n_frames // 300)   # downsample to ~300 frames for speed

def update(frame):
    i = frame * step
    t_i = t[i]

    ax.cla()

    fdx1, fdx2 = get_field(t_i)
    ax.streamplot(xG, dxG, fdx1, fdx2, color='dimgray', linewidth=0.8, density=1.2)
    ax.scatter(orbs[:, 0, i], orbs[:, 1, i], c='tomato', s=20, zorder=3)

    ax.set_xlim(xG.min(), xG.max())
    ax.set_ylim(dxG.min(), dxG.max())
    ax.set_xlabel(r'$x$')
    ax.set_ylabel(r'$\dot x$')
    ax.set_title(rf'Van der Pol (forced, $A={Af_val}$, $\omega={wf_val}$)  $t = {t_i:.2f}$')

    return []

ani = FuncAnimation(fig, update, frames=n_frames // step,
                    interval=30, blit=False)

plt.tight_layout()
HTML(ani.to_jshtml())
```
